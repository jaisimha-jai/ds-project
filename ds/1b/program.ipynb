```python
#p1
import pandas as pd
student_data={
    "Roll_No":[101,102,103,104],
    "Name":["jai","sai","navaneeth","Swayam"],
    "Department":["IT","IT","CSE","ECE"],
    "Marks":[92,90,87,86]
    }
df=pd.DataFrame(student_data)
df.to_json(
    "students.json",
    orient="records",
    indent=4
)
print("JSON file created successfully")
```

```python
#P2
import pandas as pd
df=pd.read_json("students.json")
print("Students data:")
print(df)
```

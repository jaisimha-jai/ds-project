# 1. CSV Reading Example

```python
import pandas as pd
csv_url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
csv_df = pd.read_csv(csv_url)
print("CSV Data:")
print(csv_df.head())
csv_df.to_csv("iris_output.csv", index=False)
print("\nCSV file written successfully")
```

# 2. Tab-Delimited File

```python
import pandas as pd
student_data = {
    "Roll_No": [101, 102, 103, 104],
    "Name": ["Anusha", "Bob", "Cherry", "Deepika"],
    "Department": ["IT", "IT", "CSE", "ECE"],
    "Percentage": [89, 92, 88, 85]
}
student_df = pd.DataFrame(student_data)
student_df.to_csv("student_output.txt", sep="\t", index=False)
tab_df = pd.read_csv("student_output.txt", sep="\t")
print("Tab-Delimited Data:")
print(tab_df)
```

# 3. JSON Reading and Writing

```python

student_df.to_json("student_output.json", orient="records")

json_df = pd.read_json("student_output.json")

print("JSON Data:")
print(json_df)

json_string = """
[
    {"Roll_No": 201, "Name": "Esha", "Percentage": 91},
    {"Roll_No": 202, "Name": "Farah", "Percentage": 87}
]
"""

parsed_json_df = pd.read_json(StringIO(json_string))

print("\nParsed JSON String:")
print(parsed_json_df)
```

# 4. Excel Reading and Writing

```python

student_df = pd.DataFrame(student_data)

student_df.to_excel("students.xlsx", sheet_name="Student Details", index=False)

excel_df = pd.read_excel("students.xlsx", sheet_name="Student Details")

print("Excel Data:")
print(excel_df)

print("\nAll file operations completed successfully")
```

```python
#P1
import pandas as pd
student_data={
    "Roll_No":[101,102,103,104],
    "Name":["jai","navaneeth","Sai","swayam"],
    "Department":["IT","IT","CSE","ECE"],
    "Percentage":[90,89,80,79],
}
df=pd.DataFrame(student_data)
print(df)
```

```python
#P2
df = pd.DataFrame(student_data)
df.to_excel("students.xlsx", sheet_name="Student Details", index=False)
print("Data successfully written to students.xlsx")
df = pd.read_excel("students.xlsx", sheet_name="Student Details")

print("Data read from Excel:")
print(df)
```

```python
course_data = {
    "Course_ID": ["C101", "C102", "C103"],
    "Course_Name": ["Python", "Data Science", "ML"],
    "Credits": [4, 3, 4]
}

students_df = pd.DataFrame(student_data)
courses_df = pd.DataFrame(course_data)

with pd.ExcelWriter("college_data.xlsx", engine="openpyxl") as writer:
    students_df.to_excel(writer, sheet_name="Students", index=False)
    courses_df.to_excel(writer, sheet_name="Courses", index=False)

print("Multiple sheets written successfully written to college_data.xlsx")
```

```python
excel_data = pd.read_excel("college_data.xlsx", sheet_name=None)

print("Available sheets:")
print(excel_data.keys())
print("\nStudents Sheet:")
print(excel_data["Students"])
print("\nCourses Sheet:")
print(excel_data["Courses"])
```
